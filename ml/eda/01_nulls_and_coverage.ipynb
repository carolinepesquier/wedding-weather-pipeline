```python
# Libraries:
from google.cloud import bigquery
import pandas as pd
```

```python
# Initiating BigQuery client:
client = bigquery.Client(project="wedding-weather-496115")
```

```python
# Define sources list:
sources = ['meteostat', 'open_meteo', 'visual_crossing']

# Define dictionary to store dataframes for each source:
dfs = {}

# Retrieve dataframes from BigQuery for each source:
for source in sources:
    query = f""" SELECT *
                FROM `wedding-weather-496115.silver.{source}_hourly_silver`
    """
    df_source = client.query(query).to_dataframe()
    print(df_source.head())
    dfs[source] = df_source
```

```python
# Assign a dataframe to each source:
df_meteostat = dfs['meteostat']
df_open_meteo = dfs['open_meteo']
df_visual_crossing = dfs['visual_crossing']

# Create time serie properly for each source:
for source in sources:
    df = dfs[source].set_index('datetime')
    df = df.asfreq('h')
    df = df.sort_index()
    print(f"{source}:", df.index.dtype)
    print(df.head())

# Check general info:
for source in sources:
    df = dfs[source]
    print(f"\n{source}:", df.info())
```
